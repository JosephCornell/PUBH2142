## Homework 1.3.2.1: Is 1003 prime?

Determine whether 1003 is a prime number. A number is prime if it has exactly two positive factors: 1 and itself. To answer this, we can test whether any integer from 2 through the square root of 1003 divides it evenly.

```python
import math

number = 1003
factors = [divisor for divisor in range(2, math.isqrt(number) + 1) if number % divisor == 0]

if factors:
    print(f"{number} is not prime; it is divisible by {factors[0]}.")
else:
    print(f"{number} is prime.")
```

## Homework 1.3.2.2: Prime factors of 23558

Determine the prime factorization of 23558 and the multiplicity of each prime factor. We repeatedly divide the number by its smallest available prime factor until the remaining quotient is 1. The number of times each prime divides the original number is its multiplicity.

```python
number = 23558
remaining = number
factor_multiplicities = {}
divisor = 2

while divisor * divisor <= remaining:
    while remaining % divisor == 0:
        factor_multiplicities[divisor] = factor_multiplicities.get(divisor, 0) + 1
        remaining //= divisor
    divisor += 1

if remaining > 1:
    factor_multiplicities[remaining] = factor_multiplicities.get(remaining, 0) + 1

print("Prime factors and multiplicities:")
for prime, multiplicity in factor_multiplicities.items():
    print(f"{prime}: {multiplicity}")

factorization = " * ".join(
    f"{prime}^{multiplicity}" if multiplicity > 1 else str(prime)
    for prime, multiplicity in factor_multiplicities.items()
)
print(f"{number} = {factorization}")
```

Calculate the cube root of 100000 using three approaches: pure Python with the exponentiation operator `**`, the `math` package, and the `sympy` package.
```python
constant = 5
right_side = 3
boundary = right_side - constant

print(f"x + {constant} >= {right_side}")
print(f"After subtracting {constant}: x >= {boundary}")
print(f"Interval notation: [{boundary}, infinity)")
print(f"Check boundary: {boundary} + {constant} >= {right_side} is {boundary + constant >= right_side}")
print(f"Check a value in the interval: 0 + {constant} >= {right_side} is {constant >= right_side}")
```

## Homework 9.4.2: Solving a Quadratic Inequality

Solve the inequality \(x^2 - x - 6 > 0\).

Factor the quadratic:

\[
x^2 - x - 6 = (x-3)(x+2).
\]

The critical values are \(x=3\) and \(x=-2\). They divide the number line into three intervals:

\[
(-\infty,-2), \qquad (-2,3), \qquad (3,\infty).
\]

Test the sign of the product on each interval:

- If \(x<-2\), both factors are negative, so their product is positive.
- If \(-2<x<3\), the factors have opposite signs, so their product is negative.
- If \(x>3\), both factors are positive, so their product is positive.

Because the inequality is strict, the boundary points \(-2\) and \(3\) are not included. Therefore, the solution interval is

\[
\boxed{(-\infty,-2)\cup(3,\infty)}.
\]


```python
def expression(value):
    return value**2 - value - 6

critical_values = [-2, 3]
test_values = [-3, 0, 4]

print("Factored form: (x - 3)(x + 2) > 0")
print(f"Critical values: {critical_values}")
for value in test_values:
    print(f"x = {value}: x^2 - x - 6 = {expression(value)}; positive? {expression(value) > 0}")

print("Solution interval: (-infinity, -2) union (3, infinity)")
```

## Homework 9.4.3: Solving a Rational Inequality

Solve the inequality

\frac{x^2-1}{x-3} \ge 0.

Factor the numerator:

\[
\frac{x^2-1}{x-3} = \frac{(x-1)(x+1)}{x-3}.
\]

The critical values are \(x=-1\) and \(x=1\), where the numerator is zero, and \(x=3\), where the denominator is zero. The denominator value \(x=3\) must be excluded.

Check the sign on each interval:

- On \((-1,1)\), the expression is positive.
- On \((1,3)\), the expression is negative.
- On \((3,\infty)\), the expression is positive.

Because the inequality includes equality, \(-1\) and \(1\) are included. The denominator zero \(3\) is excluded. Therefore, the solution is

\[
\boxed{[-1,1]\cup(3,\infty)}.

```python
def rational_expression(value):
    return (value**2 - 1) / (value - 3)

critical_values = [-1, 1, 3]
test_values = [-2, 0, 2, 4]

print("Factored form: (x - 1)(x + 1) / (x - 3) >= 0")
print(f"Critical values: {critical_values}")
for value in test_values:
    result = rational_expression(value)
    print(f"x = {value}: value = {result}; satisfies? {result >= 0}")

print("x = 3 is excluded because it makes the denominator zero.")
print("Solution interval: [-1, 1] union (3, infinity)")
```

## Homework 9.6.1: Solving an Absolute-Value Equation

Solve

\[
|x^3|=27.
\]

For real numbers, \(|x^3|=|x|^3\), so

\[
|x|^3=27 \quad \Longrightarrow \quad |x|=3.
\]

Thus, the real solutions are

\[
x=3 \quad \text{or} \quad x=-3.
\]

To list the complex algebraic roots suggested by the hint, split the absolute-value equation into

\[
x^3=27 \quad \text{or} \quad x^3=-27.
\]

The cube roots of \(27\) are

\[
3, \quad 3e^{2\pi i/3}, \quad 3e^{4\pi i/3},
\]

and the cube roots of \(-27\) are

\[
3e^{i\pi/3}, \quad 3e^{i\pi}, \quad 3e^{5\pi i/3}.
\]

Equivalently, the six algebraic roots are

\[
\boxed{3,\ -3,\ \frac{3}{2}+\frac{3\sqrt{3}}{2}i,\ \frac{3}{2}-\frac{3\sqrt{3}}{2}i,\ -\frac{3}{2}+\frac{3\sqrt{3}}{2}i,\ -\frac{3}{2}-\frac{3\sqrt{3}}{2}i}.
\]

Strictly speaking, if \(|x^3|\) means the complex modulus, then \(|x^3|=|x|^3\), so every complex number \(x=3e^{i\theta}\) is a solution. The six values above are the finite set obtained by the worksheet's algebraic split into \(x^3=\pm27\).


```python
from cmath import exp
from math import pi

def cube_roots(target):
    starting_angle = 0 if target > 0 else pi
    return [3 * exp(1j * (starting_angle + 2 * pi * k) / 3) for k in range(3)]

algebraic_roots = cube_roots(27) + cube_roots(-27)

print("Real solutions from |x| = 3: x = 3 and x = -3")
print("Six algebraic roots from x^3 = 27 or x^3 = -27:")
for root in algebraic_roots:
    print(f"x = {root.real:.6f} {root.imag:+.6f}j; |x^3| = {abs(root**3):.6f}")

print("Literal complex-modulus form: x = 3 * exp(i * theta), where theta is any real number.")
```

## Homework 10.4.2: Equation of a Perpendicular Line

Find the equation of the line perpendicular to \(y=3x-1\) that passes through \((1,2)\).

The given line has slope \(m=3\). Perpendicular nonvertical lines have slopes that are negative reciprocals, so the perpendicular slope is

\[
m_{\perp}=-\frac{1}{3}.
\]

Use point-slope form with the point \((1,2)\):

\[
y-2=-\frac{1}{3}(x-1).
\]

Simplify:

\[
y-2=-\frac{1}{3}x+\frac{1}{3}
\]

\[
\boxed{y=-\frac{1}{3}x+\frac{7}{3}}.
\]

The slopes satisfy \(3\left(-\frac{1}{3}\right)=-1\), confirming that the two lines are perpendicular.


```python
from fractions import Fraction

original_slope = Fraction(3)
perpendicular_slope = -1 / original_slope
x_point = 1
y_point = 2
intercept = y_point - perpendicular_slope * x_point

print(f"Original slope: {original_slope}")
print(f"Perpendicular slope: {perpendicular_slope}")
print(f"Equation: y = ({perpendicular_slope})x + ({intercept})")
print(f"Passes through (1, 2): {perpendicular_slope * x_point + intercept == y_point}")
print(f"Product of slopes: {original_slope * perpendicular_slope}")
```

## Homework 10.4.4: Writing a Line Through Two Points

Given the points \((1,0)\) and \((-3,4)\), write the equation of the line in slope-intercept form.

First find the slope:

\[
m=\frac{4-0}{-3-1}=\frac{4}{-4}=-1.
\]

Use \(y=mx+b\) and substitute the point \((1,0)\):

\[
0=-1(1)+b.
\]

Therefore, \(b=1\), and the equation is

\[
\boxed{y=-x+1}.
\]

Checking the second point gives \(y=-(-3)+1=4\), so both points lie on the line.


```python
from fractions import Fraction

first_point = (1, 0)
second_point = (-3, 4)
slope = Fraction(second_point[1] - first_point[1], second_point[0] - first_point[0])
intercept = first_point[1] - slope * first_point[0]

def line_value(x_value):
    return slope * x_value + intercept

print(f"Slope: {slope}")
print(f"Equation: y = {slope}x + ({intercept})")
print(f"At {first_point}: y = {line_value(first_point[0])}")
print(f"At {second_point}: y = {line_value(second_point[0])}")
print(f"Both points verified: {line_value(first_point[0]) == first_point[1] and line_value(second_point[0]) == second_point[1]}")
```
